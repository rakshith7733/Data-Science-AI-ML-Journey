```python
Dictionaries: A dictionary in python functions similarly to a dictornary in real life. It's a data structure that allows us to associate a key to a value and pair the two pieces of data together.
- Defined by {}

Syntax:

Dict = {"apple": "red",
        "banana": "yellow",
        "pear": "green"
        }
O/P: {'apple': 'red', 'banana': 'yellow', 'pear': 'green'}


As like any other data structure for creating empty variable we can use as below:
string_variable = ""
list_variable = []
dictionary_variable = {}

Keyerror: If a key doesnt exist in the dictornary you will get a keyerror.
```

````python
#Dictionary creating
Dict = {"apple": "red",
        "banana": "yellow",
        "pear": "green"
        }
print(Dict)
print("----------------------access a key--------------------")

#to access a key of the dictionary
print(Dict["banana"])
print("-----------------------add new entry------------------")

#if you need to add new entry to the dictionary.
Dict["strawberry"] = "pink"
print(Dict)
print("----------------------loop through dictionary----------")

#Loop through a dictionary.
for thing in Dict:
    print(thing)
#this will only produce the key elements in the dictionary
print("-------------------------------------------------------")

#to loop key and values.
for key in Dict:
    print(key)
    print(Dict[key])
````

````python
#To check if a student scores based on criteria.
students_scores = {
    "harry":88,
    "Ron":78,
    "Hermione":95,
    "Daraco": 75,
    "Neville":60
}

#create a empty dictionary
student_grades = {}

for student in students_scores:
    score = students_scores[student]
    #print(score)

    if score >=91:
        student_grades[student] = "Outstanding"
    elif score >=81:
        student_grades[student] = "Exceeds Expectations"
    elif score >=71:
        student_grades[student] = "Acceptable"
    else:
        student_grades[student] = "Fail"
print(student_grades)

````
