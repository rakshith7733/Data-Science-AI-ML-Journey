#### $$ 08 - I/O-with-Basics-files $$

File handling allows to read, write, & store data into files.


| Mode   | Meaning           |
| ------ | ----------------- |
| `"r"`  | Read              |
| `"w"`  | Write (overwrite) |
| `"a"`  | Append            |
| `"x"`  | Create new file   |
| `"rb"` | Read binary       |
| `"wb"` | Write binary      |



````python
#Opening a File.
my_file = open('Plain text.txt',"r")
````

##### Reading Files

````python
#read file
my_file.read()
````

````python
#close a file
my_file.close()
````

````python
#readlines - to read one line
print(my_file.readline())
````

```python
if you have a file in desginated location and you want to open the file, you can use the path and assign to a variable.
```

````python
#Create file
open("new.txt", "x")
````

````python
with open("Plain text.txt", "r") as f:
    print(f.read())
````

````python
#writing to a file
with open("Plain text.txt","a") as f:
    f.write("This is a new write line")
````

````python
#read file again to see the changes
with open("Plain text.txt", "r") as f:
    print(f.read())
````

##### Writing Files

````python
#Write Mode
my_file = open('Plain text.txt',"w") #this will over write the existing record in the file.
my_file.write("\nAdd this text as well") 
#It will delete the existing record inside in the files.
my_file.close()
````
