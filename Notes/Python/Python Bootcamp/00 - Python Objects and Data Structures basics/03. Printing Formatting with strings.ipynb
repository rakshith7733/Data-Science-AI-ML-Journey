#### $$ 03 - Printing-Formatting-with-strings $$

```python
String fromating help you concatenate into your string for printing.

player = 'Rakshith'
points = 05

'Last night, '+ player +' scored '+ str(points) +' points.'  # concatenation

f'Last night, {player} scored {points} points.'          # string formatting
```

````python
x, y = 'some', 'more'
print("I'm going to inject %s text here, and %s text here."%(x,y))
````

They are two methods:
1. .format() method
2. f-strings(formatted string literals)

````python
#.format() usage using {}
print("this is a string {}".format('used'))
````

````python
print("The {} is a {} place to {}.".format("village", "better", "live"))
````

````python
#using index values of format objects
print("The {1} is a {0} place to {2}.".format("village", "better", "live"))
````

````python
#usage of variables in format objects
print("The {v} is a {b} place to {l}.".format(v="village",b= "better",l= "live"))
````

```python
Float formatting follows "result: {value:{width}.{precision}}"
Where with the .format() method you might see {value:10.4f}, with f-strings this can become {value:{10}.{6}}
```

````python
print("the result was {r:1.4f}".format(r=125.673845))
````

````python
r = "result"
print(f"This is a good {r}.")
````
