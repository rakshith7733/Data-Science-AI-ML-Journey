```python
Numbers: These are whole number, which is having Integer datatype.
         Ex: 1,44,99

Strings: These are ordered sequence of characters, which is having Str datatype.
        Ex: "Hello"

Lists: These are ordered collection of elements, defined by [] brackets, mutable, can store different datatypes.
        Ex: [1,"HELLO",4,5,"good"]

Tuples: these are collection of elements like lists defined by (), immutable, cannot assign values once created.
       Ex: (1,2,"hello",5)

Dictionaries: These are key value pair combination that is unordered., defined by {}, {key:value} - for good lookup using key.
    Ex: {"key1": "value1",
         "key2": "value2"
        }
```

Using keys and indexing, grab the 'hello' from the following dictionaries:

````python
d = {'simple_key':'hello'}
# Grab 'hello'
print(d['simple_key'])
````

````python
d = {'k1':{'k2':'hello'}}
# Grab 'hello
print(d['k1']['k2'])
````

````python
# Getting a little tricker
d = {'k1':[{'nest_key':['this is deep',['hello']]}]}
print(d['k1'][0]['nest_key'][1][0])
````

````python
#Grab hello
# This will be hard and annoying!
d = {'k1':[1,2,{'k2':['this is tricky',{'tough':[1,2,['hello']]}]}]}
print(d['k1'][2]['k2'][1]['tough'][2][0])
````
