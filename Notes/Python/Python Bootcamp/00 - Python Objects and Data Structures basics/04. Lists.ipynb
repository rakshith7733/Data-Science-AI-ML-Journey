````python
# Assign a list to an variable named my_list
my_list = [1,2,3]
````

````python
#We just created a list of integers, but lists can actually hold different object types. For example:
my_list = ['string',23,100.232,'o']
````

````python
#Just like strings, the len() function will tell you how many items are in the sequence of the list.
len(my_list)
````

#### ```Indexing & Slicing```

````python
my_list = ['one',"two","three"]

#accessing element value 1
my_list[1]
````

````python
#accessing element from 1 to 3 from list.
my_list[1:3]
````

````python
# Reassign using concatenation
my_list = my_list + ['four']
my_list
````

````python
#changing the character of the elements
my_list[0] = "zero"
my_list
````

```python
We can use methods like

pop() - to remove last element from the list. if we give index position to be removed then we can give inside pop function
Ex: list.pop()
    list.pop(2)


sort() - This will sort the list in ASC order

reverse() - this will reverse the list.

and so on.....
```

````python
#removing last element from the list.
my_list.pop()
````

````python
#sorting the list.
my_list.sort()
print(my_list)
````

````python
#reverse the list
my_list.reverse()
print(my_list)
````

````python
#Knowledge:

#To index a nested list we need to add another set of brackets to find exact position of the value we needed.

set_list = [1,2,3,[1,2,3]]

#I want to pick the element of 2 in the nested list.
print(set_list[3][1])
````
