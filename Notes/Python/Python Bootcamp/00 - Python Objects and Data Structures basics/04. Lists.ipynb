```python
Lists are ordered sequence that can hold variety of objects types.
- defined by [] and commas to seperate the objects in the list.

Ex: [1,2,"Hello",4, "go"]

- list supports indexing and slicing. Lists can be nested and can be called using variety of methods.

1.) Creating lists
2.) Indexing and Slicing Lists
3.) Basic List Methods
4.) Nesting Lists
5.) Introduction to List Comprehensions
```

````python
# Assign a list to an variable named my_list
my_list = [1,2,3]
````

````python
#We just created a list of integers, but lists can actually hold different object types. For example:
my_list = ['string',23,100.232,'o']
````

````python
#Just like strings, the len() function will tell you how many items are in the sequence of the list.
len(my_list)
````

#### ```Indexing & Slicing```

````python
my_list = ['one',"two","three"]

#accessing element value 1
my_list[1]
````

````python
#accessing element from 1 to 3 from list.
my_list[1:3]
````
