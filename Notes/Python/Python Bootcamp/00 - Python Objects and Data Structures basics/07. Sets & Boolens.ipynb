#### $$ 07 - Sets-and-Boolens $$

```python
Sets are unordered collections of unique elements.
- Unordered
- Unique values only
- Mutable
- defined using {}

Sets are fast:o(1) average.
- Lookup, insert, remove

when to use:
- need unique values.
- removing duplicates
- fast membership checking
- comparing collections
```

````python
#creating a set
numbers = {1,2,3,4}
print(numbers)
````

````python
#Duplicates values automatically removed
data = {1,2,2,3,3,1,4}
print(data)
````

````python
#Adding Elements
data.add(5)
print(data)
````

````python
#remove element
data.remove(5)
print(data) #gets error if the element does't present in the tuple.
````
