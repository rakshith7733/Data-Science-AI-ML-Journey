#### $$ 05 - Dicitionary $$

```python
Dictionary are unordered mappings for storing objects, they use key:value pair.
- Key:value pair helping grab the objects quickly without need of index position.

Syntax: {"key1":"value1",
         "key2":"value2,
         "key3":"value3"
         }

properties:
- Mutable
- key value structure fast lookup
- keys must be unique.


When to choose?
- Objects retrieved by key name.
- Data has labels.
- Real world structured objects.

Good for:
- User profiles
- JSON/API Data
- ML configurations
- Database like records.
```

````python
#Creating a Dictionary

student = {
    "name": "Rakshith",
    "age": 25,
    "course": "Python"
}

#Accessing Values
print(student["name"])
````

````python
#Adding New Data
student["city"] = "Hyderabad"
print(student)
````

````python
#Updating Values
student["age"] = 26
print(student)
````

````python
#Removing Data
del student["course"]
print(student)
````

````python
#Dictionary methods
print("this to fetch keys: ",student.keys())

student.values()
print("this to fetch values: ",student.values())

student.items()
print("this to fetch items: ",student.items())

#safer way to access values.
student.get("name")
````

````python
#Combination of list and nested dictionary in a dictionary.
mix = {"k1":123,
       "k2":[0,1,2,3],
       "k3":{'insidekey':100}
       }


#Access 'insidekey' value from a dictionary.
print(mix["k3"]['insidekey'])
````
