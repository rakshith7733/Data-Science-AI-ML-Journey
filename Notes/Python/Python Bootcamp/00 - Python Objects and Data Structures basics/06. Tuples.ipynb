#### $$ 06 - Tuples $$

```python
Tuples are collection of items as like list.
- These are Immutable.
- Once an element assigned cannot be reassigned.
- Defined by parenthesis ()

Ex: Tup = (1,2,3,"good","bad")

~ Use when:
Data should NOT change
Fixed values
Safer data handling
Better performance
Dictionary keys
```

````python
#Creating a Tuple
numbers = (1, 2, 3, 4,3)

#Accessing Elements - Uses indexing like lists.
print(numbers[0])
````

````python
numbers[0] = 10
````

````python
#tuple methods
numbers.count(3)
````

````python
#index of the item.
numbers.index(2)
````

````python
#packing
data = (10,20,30)

#unpacking
a,b,c = data
print(a)
print(b)
print(c)
````

````python
#list inside tuple
l_tup = (1,2,[1,2,3])
print(l_tup)
````

````python
l_tup[2].append(5)
print(l_tup)

#here we can modify the list due to its mutability but not the tuple.
````

````python
x = ([1,2], [3,4])
x[0].append(10)
print(x)
````
